```python
import numpy as np

# --- PARAMETERS ---
LOW = 670
HIGH = 920
N_PLAYERS = 10000
N_SIM = 10000

# your choice
my_b2_choices = [914, 915, 916, 917, 918, 919]

# assume competitors' behavior (constantly change)
def generate_competitors(n):
    # clustered near high values
    top_cluster = np.random.normal(loc=917.5, scale=2, size=int(0.7 * n)) # loc center to change
    rest = np.random.uniform(916, 919, size=int(0.3 * n))
    bids = np.concatenate([top_cluster, rest])
    return np.clip(bids, 900, 920)

def payoff(my_b2, avg_b2):
    base = 920 - my_b2
    if my_b2 > avg_b2:
        return base
    else:
        penalty = ((920 - avg_b2) / (920 - my_b2)) ** 3
        return base * penalty

# --- SIMULATION ---
results = {b2: [] for b2 in my_b2_choices}

for _ in range(N_SIM):
    competitors = generate_competitors(N_PLAYERS)
    avg_b2 = np.mean(competitors)

    for b2 in my_b2_choices:
        results[b2].append(payoff(b2, avg_b2))

# --- OUTPUT ---
for b2 in my_b2_choices:
    avg_profit = np.mean(results[b2])
    print(f"b2 = {b2}: expected profit ≈ {avg_profit:.2f}")
```
