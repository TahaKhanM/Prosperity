```python
import numpy as np
import matplotlib.pyplot as plt

# Constants
LOW = 670
HIGH = 920
STEP = 5
N = (HIGH - LOW) // STEP + 1  # 51 points

# Discrete EV function
def discrete_ev(b1):
    k = 1 if b1 > LOW else 0
    k += (b1 - 1 - LOW) // STEP # number of valid reserve prices ≤ b1
    return k * (HIGH - b1) # number of valid reserve prices times profit

# # Continuous EV function (approximation)
# def continuous_ev(b1):
#     return ((b1 - LOW) / (HIGH - LOW)) * (HIGH - b1)

# Domain (try over all valid bids)
bids = np.arange(LOW, HIGH + 1, 1)

# Compute EVs
ev_discrete = np.array([discrete_ev(b) for b in bids])
# ev_continuous = continuous_ev(bids)

# Find optimal discrete bid
opt_idx = np.argmax(ev_discrete)
opt_bid = bids[opt_idx]
opt_value = ev_discrete[opt_idx]

# Plot
plt.figure(figsize=(10, 6))

plt.plot(bids, ev_discrete, marker='o', label='Discrete EV (true)', linewidth=2)
# plt.plot(bids, ev_continuous, linestyle='--', label='Continuous Approximation')

# Highlight optimum
plt.scatter(opt_bid, opt_value, color='red', s=100, zorder=5, label=f'Optimal b1 = {opt_bid}')

plt.title("Expected Value of First Bid")
plt.xlabel("First Bid (b1)")
plt.ylabel("Expected Profit")
plt.legend()
plt.grid(True)

plt.show()

# Print result
print(f"Optimal discrete b1: {opt_bid}")
print(f"Expected profit at optimum: {opt_value:.2f}")



```
