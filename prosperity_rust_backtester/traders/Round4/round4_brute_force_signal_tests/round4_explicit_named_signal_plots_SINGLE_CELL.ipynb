# Round 4 explicit counterparty signal plots

Single-cell notebook that directly plots the named Mark→Mark signals, plus tables and heatmaps.

```python

# Round 4: explicit graph review for the named counterparty signals
# Everything is in this one cell. Put the CSVs in this notebook folder or set DATA_DIR manually.

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path
import re
import os

# =====================
# 0) CONFIG
# =====================
# If auto-detect fails, set this to the folder containing prices_round_4_day_*.csv and trades_round_4_day_*.csv
DATA_DIR = '/Users/andys/Shared Claude Code/Prosperity/prosperity_rust_backtester/traders/Round4/Round4 Brute Force Signal Tests'  # example: Path("/Users/andy/Shared Claude Code/Prosperity")

HORIZONS = [100, 200, 500, 1000, 2000, 5000, 10000]

NAMED_SIGNALS = [
    # product, buyer, seller, direction, horizon_to_plot, note
    ("VELVETFRUIT_EXTRACT", "Mark 55", "Mark 14", "LONG_AFTER_TRADE", 10000, "Mark 55 buys from Mark 14 -> LONG VFE"),
    ("VELVETFRUIT_EXTRACT", "Mark 14", "Mark 55", "SHORT_AFTER_TRADE", 10000, "Mark 14 buys from Mark 55 -> SHORT VFE"),
    ("VELVETFRUIT_EXTRACT", "Mark 67", "Mark 49", "LONG_AFTER_TRADE", 1000, "Mark 67 buys from Mark 49 -> LONG VFE"),
    ("VELVETFRUIT_EXTRACT", "Mark 67", "Mark 22", "LONG_AFTER_TRADE", 1000, "Mark 67 buys from Mark 22 -> LONG VFE"),
    ("HYDROGEL_PACK", "Mark 38", "Mark 14", "SHORT_AFTER_TRADE", 5000, "Mark 38 buys from Mark 14 -> SHORT HYDROGEL"),
    ("HYDROGEL_PACK", "Mark 14", "Mark 38", "LONG_AFTER_TRADE", 5000, "Mark 14 buys from Mark 38 -> LONG HYDROGEL; weaker / not clean on day 3"),
]

# =====================
# 1) FIND FILES
# =====================
def find_data_dir():
    if DATA_DIR is not None:
        return Path(DATA_DIR).expanduser().resolve()
    candidates = []
    cwd = Path.cwd().resolve()
    candidates += [cwd, cwd.parent, cwd.parent.parent]
    candidates += [Path("/mnt/data")]
    # Add common user home search, shallow only
    home = Path.home()
    candidates += [home / "Shared Claude Code" / "Prosperity", home / "Prosperity", home / "Downloads"]
    seen = set()
    for c in candidates:
        c = c.resolve()
        if c in seen:
            continue
        seen.add(c)
        if list(c.glob("prices_round_4_day_*.csv")) and list(c.glob("trades_round_4_day_*.csv")):
            return c
    print("Could not auto-detect CSV folder. Looked in:")
    for c in seen:
        print("  ", c)
    raise FileNotFoundError("Set DATA_DIR at the top of this cell to your CSV folder.")

DATA_DIR_FOUND = find_data_dir()
PRICE_FILES = sorted(DATA_DIR_FOUND.glob("prices_round_4_day_*.csv"))
TRADE_FILES = sorted(DATA_DIR_FOUND.glob("trades_round_4_day_*.csv"))
print("Using DATA_DIR:", DATA_DIR_FOUND)
print("Price files:", [f.name for f in PRICE_FILES])
print("Trade files:", [f.name for f in TRADE_FILES])

# =====================
# 2) LOAD DATA
# =====================
def day_from_path(path):
    m = re.search(r"day_(\d+)", str(path))
    if not m:
        raise ValueError(f"Could not parse day from filename: {path}")
    return int(m.group(1))

prices = []
for f in PRICE_FILES:
    df = pd.read_csv(f, sep=";")
    if "day" not in df.columns:
        df["day"] = day_from_path(f)
    prices.append(df)
prices = pd.concat(prices, ignore_index=True)

trades = []
for f in TRADE_FILES:
    df = pd.read_csv(f, sep=";")
    if "day" not in df.columns:
        df["day"] = day_from_path(f)
    trades.append(df)
trades = pd.concat(trades, ignore_index=True).rename(columns={"symbol": "product"})

# Normalize whitespace just in case.
for col in ["buyer", "seller", "product"]:
    if col in trades.columns:
        trades[col] = trades[col].astype(str).str.strip()
prices["product"] = prices["product"].astype(str).str.strip()

print("\nData loaded")
print("prices:", prices.shape)
print("trades:", trades.shape)
print("products:", sorted(prices["product"].unique()))
print("marks:", sorted(set(trades["buyer"]).union(set(trades["seller"]))))

# =====================
# 3) ATTACH FORWARD RETURNS TO EVERY TRADE
# =====================
mid = prices[["day", "timestamp", "product", "mid_price"]].copy()
events = trades.merge(
    mid.rename(columns={"mid_price": "mid_t"}),
    on=["day", "timestamp", "product"],
    how="left",
)

for H in HORIZONS:
    future_mid = mid.copy()
    future_mid["timestamp"] -= H  # so row t joins to price at t+H
    events = events.merge(
        future_mid.rename(columns={"mid_price": f"mid_fwd_{H}"}),
        on=["day", "timestamp", "product"],
        how="left",
    )
    events[f"ret_{H}"] = events[f"mid_fwd_{H}"] - events["mid_t"]
    events[f"qty_ret_{H}"] = events["quantity"] * events[f"ret_{H}"]

# =====================
# 4) TABLE: EXACT STATS FOR THE NAMED SIGNALS
# =====================
def signal_rows_for_all_horizons(product, buyer, seller, direction):
    out = []
    sub0 = events[(events["product"] == product) & (events["buyer"] == buyer) & (events["seller"] == seller)].copy()
    sign = 1 if direction == "LONG_AFTER_TRADE" else -1
    for H in HORIZONS:
        sub = sub0[sub0[f"ret_{H}"].notna()].copy()
        if len(sub) == 0:
            out.append({"product": product, "buyer": buyer, "seller": seller, "direction": direction, "horizon": H, "events": 0, "days": 0, "score": np.nan, "avg_per_event": np.nan, "worst_day": np.nan})
            continue
        sub["points"] = sign * sub["quantity"] * sub[f"ret_{H}"]
        daily = sub.groupby("day")["points"].sum()
        out.append({
            "product": product,
            "buyer": buyer,
            "seller": seller,
            "direction": direction,
            "horizon": H,
            "events": len(sub),
            "days": sub["day"].nunique(),
            "score": sub["points"].sum(),
            "avg_per_event": sub["points"].mean(),
            "worst_day": daily.min(),
        })
    return out

rows = []
for product, buyer, seller, direction, plot_horizon, note in NAMED_SIGNALS:
    rows += signal_rows_for_all_horizons(product, buyer, seller, direction)
summary = pd.DataFrame(rows)
summary["score"] = summary["score"].round(2)
summary["avg_per_event"] = summary["avg_per_event"].round(3)
summary["worst_day"] = summary["worst_day"].round(2)

print("\n=== Exact named-signal score table across horizons ===")
display(summary.sort_values(["product", "buyer", "seller", "direction", "horizon"]))

# Best horizon per named signal, by total score.
best = summary.dropna().sort_values("score", ascending=False).groupby(["product", "buyer", "seller", "direction"], as_index=False).head(1)
print("\n=== Best horizon for each named signal ===")
display(best[["product", "buyer", "seller", "direction", "horizon", "events", "days", "score", "avg_per_event", "worst_day"]])

# =====================
# 5) PLOTTING HELPERS
# =====================
def matching_events(product, buyer, seller):
    return events[(events["product"] == product) & (events["buyer"] == buyer) & (events["seller"] == seller)].copy()

def plot_signal(product, buyer, seller, direction="LONG_AFTER_TRADE", horizon=1000, day=None, show_trade_price=True):
    sig = matching_events(product, buyer, seller)
    if day is not None:
        sig = sig[sig["day"] == day]
    days = sorted(sig["day"].unique())
    if not days:
        print(f"No matching events for {product}: {buyer} -> {seller}")
        return
    sign = 1 if direction == "LONG_AFTER_TRADE" else -1
    arrow = "↑ good" if direction == "LONG_AFTER_TRADE" else "↓ good"

    for d in days:
        p = prices[(prices["product"] == product) & (prices["day"] == d)].sort_values("timestamp")
        s = sig[(sig["day"] == d) & sig[f"ret_{horizon}"].notna()].sort_values("timestamp").copy()
        if s.empty:
            print(f"No fwd returns for day={d}, horizon={horizon}: {product} {buyer}->{seller}")
            continue
        s["event_points"] = sign * s["quantity"] * s[f"ret_{horizon}"]
        total = s["event_points"].sum()
        avg = s["event_points"].mean()

        plt.figure(figsize=(15, 5))
        plt.plot(p["timestamp"], p["mid_price"], linewidth=1, label="mid price")
        plt.scatter(s["timestamp"], s["mid_t"], s=55, marker="o", label="signal trade")
        if show_trade_price:
            plt.scatter(s["timestamp"], s["price"], s=25, marker="x", label="actual trade price")

        # Draw line from signal mid to forward mid. For long, upward line is good. For short, downward line is good.
        for _, r in s.iterrows():
            x0, x1 = r["timestamp"], r["timestamp"] + horizon
            y0, y1 = r["mid_t"], r[f"mid_fwd_{horizon}"]
            if pd.notna(y1):
                plt.plot([x0, x1], [y0, y1], linewidth=1, alpha=0.45)

        plt.title(f"{product} day {d}: {buyer} buys from {seller} => {direction} ({arrow})\n"
                  f"horizon={horizon}, events={len(s)}, score={total:.1f}, avg/event={avg:.2f}")
        plt.xlabel("timestamp")
        plt.ylabel("price")
        plt.grid(True, alpha=0.25)
        plt.legend()
        plt.show()

        plt.figure(figsize=(15, 3.8))
        plt.plot(s["timestamp"], s["event_points"].cumsum(), marker="o")
        plt.axhline(0, linewidth=1)
        plt.title(f"Cumulative event score for same signal: {product} day {d}, horizon={horizon}")
        plt.xlabel("timestamp")
        plt.ylabel("qty × signed forward mid move")
        plt.grid(True, alpha=0.25)
        plt.show()

# =====================
# 6) EXPLICITLY PLOT THE SIGNALS YOU ASKED ABOUT
# =====================
print("\n=== Explicit plots for the named signals ===")
for product, buyer, seller, direction, horizon, note in NAMED_SIGNALS:
    print("\n" + "=" * 100)
    print(note)
    print(f"Plotting: product={product}, buyer={buyer}, seller={seller}, direction={direction}, horizon={horizon}")
    plot_signal(product, buyer, seller, direction=direction, horizon=horizon)

# =====================
# 7) INDIVIDUAL BUYER -> SELLER HEATMAPS FOR CONTEXT
# =====================
def relationship_matrix(product, horizon=1000, direction="LONG_AFTER_TRADE", day=None):
    marks = sorted(set(trades["buyer"]).union(set(trades["seller"])))
    sub = events[(events["product"] == product) & events[f"ret_{horizon}"].notna()]
    if day is not None:
        sub = sub[sub["day"] == day]
    M = pd.DataFrame(0.0, index=marks, columns=marks)
    C = pd.DataFrame(0, index=marks, columns=marks)
    sign = 1 if direction == "LONG_AFTER_TRADE" else -1
    for r in sub.itertuples(index=False):
        val = sign * float(r.quantity) * float(getattr(r, f"ret_{horizon}"))
        M.loc[r.buyer, r.seller] += val
        C.loc[r.buyer, r.seller] += 1
    return M, C

def plot_relationship_heatmap(product, horizon=1000, direction="LONG_AFTER_TRADE", min_count=1):
    M, C = relationship_matrix(product, horizon, direction)
    A = M.copy()
    A[C < min_count] = np.nan
    plt.figure(figsize=(8, 6))
    plt.imshow(A.values, aspect="auto")
    plt.xticks(range(len(A.columns)), A.columns, rotation=45, ha="right")
    plt.yticks(range(len(A.index)), A.index)
    plt.colorbar(label="quantity-weighted signed forward points")
    plt.title(f"{product}: {direction}, horizon={horizon}, min_count={min_count}\ny-axis = buyer, x-axis = seller")
    plt.xlabel("seller")
    plt.ylabel("buyer")
    plt.tight_layout()
    plt.show()
    print("Score matrix:")
    display(M.round(1))
    print("Event-count matrix:")
    display(C)

print("\n=== Heatmaps for the main products/signals ===")
plot_relationship_heatmap("VELVETFRUIT_EXTRACT", horizon=1000, direction="LONG_AFTER_TRADE", min_count=3)
plot_relationship_heatmap("VELVETFRUIT_EXTRACT", horizon=10000, direction="SHORT_AFTER_TRADE", min_count=3)
plot_relationship_heatmap("HYDROGEL_PACK", horizon=5000, direction="SHORT_AFTER_TRADE", min_count=3)
plot_relationship_heatmap("HYDROGEL_PACK", horizon=5000, direction="LONG_AFTER_TRADE", min_count=3)

print("\nDone. The important thing: the listed signals are now explicitly plotted above, not hidden behind the generic top-candidate graph.")
```
